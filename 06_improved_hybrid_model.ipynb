```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import MinMaxScaler

import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Dense, Dropout, Conv1D,
    GlobalAveragePooling1D, GlobalAveragePooling2D,
    MultiHeadAttention, LayerNormalization,
    Concatenate
)
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping

# =========================
# LOAD DATA
# =========================
X_seq = np.load("../data/X_sequences.npy")
y = np.load("../data/y_targets.npy")
dates = pd.read_csv("../data/sequence_dates.csv")
dates['Date'] = pd.to_datetime(dates['Date'])

# =========================
# LOAD IMAGE DATA AGAIN
# =========================
import os
from PIL import Image

df = pd.read_csv("../data/cleaned_aqi_final.csv")
df['Date'] = pd.to_datetime(df['Date'])
df = df[df['Date'].dt.year.isin([2020, 2021, 2024])]

base_folder = "../satellite images"

image_paths = []
for root, dirs, files in os.walk(base_folder):
    for file in files:
        if file.endswith((".png", ".jpg", ".jpeg")):
            image_paths.append(os.path.join(root, file))

IMG_SIZE = (224, 224)

processed_images = []
image_dates = []

for path in image_paths:
    img = Image.open(path).convert("RGB")
    img = img.resize(IMG_SIZE)
    img = np.array(img) / 255.0

    processed_images.append(img)

    filename = os.path.basename(path)
    date_part = filename.split(".")[0]
    image_dates.append(pd.to_datetime(date_part))

processed_images = np.array(processed_images)

image_df = pd.DataFrame({
    "Date": image_dates
})

# =========================
# MATCH IMAGE + SEQUENCE DATES
# =========================
final_df = pd.merge(dates, image_df, on="Date", how="inner")

matched_images = []
matched_seq = []
matched_targets = []

date_to_index = {d: i for i, d in enumerate(dates['Date'])}
image_date_to_index = {d: i for i, d in enumerate(image_dates)}

for d in final_df['Date']:
    matched_images.append(processed_images[image_date_to_index[d]])
    matched_seq.append(X_seq[date_to_index[d]])
    matched_targets.append(y[date_to_index[d]])

X_img = np.array(matched_images)
X_seq_final = np.array(matched_seq)
y_final = np.array(matched_targets)

print("Final matched samples:", X_img.shape[0])
print(X_img.shape)
print(X_seq_final.shape)
print(y_final.shape)

# =========================
# SCALE TARGET
# =========================
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y_final.reshape(-1, 1))

# =========================
# TRAIN TEST SPLIT
# =========================
X_img_train, X_img_test, X_seq_train, X_seq_test, y_train, y_test = train_test_split(
    X_img,
    X_seq_final,
    y_scaled,
    test_size=0.2,
    random_state=42
)

# =========================
# IMAGE BRANCH
# =========================
base = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base.trainable = False

img_input = Input(shape=(224, 224, 3))
img_features = base(img_input)
img_features = GlobalAveragePooling2D()(img_features)
img_features = Dense(128, activation='relu')(img_features)

# =========================
# SEQUENCE BRANCH
# =========================
seq_input = Input(shape=(14, X_seq_final.shape[2]))

x = Conv1D(64, 3, padding='same', activation='relu')(seq_input)
x = Conv1D(64, 3, padding='same', activation='relu')(x)

attn = MultiHeadAttention(
    num_heads=4,
    key_dim=32
)(x, x)

x = x + attn
x = LayerNormalization()(x)

x = GlobalAveragePooling1D()(x)
x = Dense(128, activation='relu')(x)

# =========================
# HYBRID FUSION
# =========================
combined = Concatenate()([img_features, x])

combined = Dense(128, activation='relu')(combined)
combined = Dropout(0.3)(combined)
combined = Dense(64, activation='relu')(combined)

output = Dense(1)(combined)

model = Model(
    inputs=[img_input, seq_input],
    outputs=output
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(0.0003),
    loss='mse',
    metrics=['mae']
)

model.summary()

# =========================
# TRAIN
# =========================
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    [X_img_train, X_seq_train],
    y_train,
    validation_split=0.2,
    epochs=20,
    batch_size=16,
    callbacks=[early_stop]
)

# =========================
# EVALUATE
# =========================
pred = model.predict([X_img_test, X_seq_test])

pred_actual = scaler.inverse_transform(pred)
y_actual = scaler.inverse_transform(y_test)

mae = mean_absolute_error(y_actual, pred_actual)
rmse = np.sqrt(mean_squared_error(y_actual, pred_actual))
r2 = r2_score(y_actual, pred_actual)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)

# =========================
# PLOT
# =========================
plt.figure(figsize=(8,5))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.legend()
plt.show()
```

```python
# =========================
# ACCURACY WITH TOLERANCE
# =========================

error = np.abs(y_actual - pred_actual)

# prediction considered correct if error <= 50 AQI
correct = np.sum(error <= 50)

accuracy = (correct / len(error)) * 100

print("Model Accuracy:", accuracy, "%")
```

```python
# =========================
# AQI CATEGORY ACCURACY
# =========================

def get_category(aqi):
    if aqi <= 50:
        return 0
    elif aqi <= 100:
        return 1
    elif aqi <= 200:
        return 2
    elif aqi <= 300:
        return 3
    elif aqi <= 400:
        return 4
    else:
        return 5

actual_cat = np.array([get_category(x[0]) for x in y_actual])
pred_cat = np.array([get_category(x[0]) for x in pred_actual])

category_accuracy = np.mean(actual_cat == pred_cat) * 100

print("AQI Category Accuracy:", category_accuracy, "%")
```
