```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# load cleaned dataset
df = pd.read_csv("../data/cleaned_aqi_final.csv")

# convert date
df['Date'] = pd.to_datetime(df['Date'])

# extract year and month
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.strftime('%b')

# yearly AQI
yearly_aqi = df.groupby('Year')['AQI'].mean()

plt.figure(figsize=(8,5))
yearly_aqi.plot(kind='bar')
plt.title('Average AQI by Year')
plt.ylabel('AQI')
plt.show()

# monthly AQI
monthly_aqi = df.groupby('Month_Name')['AQI'].mean().reindex(
    ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
)

plt.figure(figsize=(10,5))
monthly_aqi.plot(kind='bar')
plt.title('Average AQI by Month')
plt.ylabel('AQI')
plt.show()
```
