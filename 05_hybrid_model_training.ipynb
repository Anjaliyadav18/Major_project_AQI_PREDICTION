```python
import pandas as pd
import numpy as np
import glob
import os
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, Flatten, Dense,
    Dropout, GlobalAveragePooling1D, MultiHeadAttention,
    LayerNormalization, Add, Conv1D, Concatenate
)
from tensorflow.keras.models import Model
import matplotlib.pyplot as plt

# =====================================
# LOAD AQI DATA
# =====================================
df = pd.read_csv("../data/cleaned_aqi_final.csv")
df['Date'] = pd.to_datetime(df['Date'])
df = df[df['Date'].dt.year == 2024].copy()

# =====================================
# LOAD IMAGE DATA
# =====================================
image_paths = glob.glob(r"..\sattelite images\**\*.png", recursive=True)
image_paths += glob.glob(r"..\sattelite images\**\*.jpg", recursive=True)
image_paths += glob.glob(r"..\sattelite images\**\*.jpeg", recursive=True)

IMG_SIZE = (224, 224)

processed_images = []
image_dates = []

for path in image_paths:
    img = Image.open(path).convert('RGB')
    img = img.resize(IMG_SIZE)
    img = np.array(img) / 255.0

    processed_images.append(img)

    filename = os.path.basename(path)
    filename = filename.replace(".png", "").replace(".jpg", "").replace(".jpeg", "")
    image_dates.append(pd.to_datetime(filename))

processed_images = np.array(processed_images)

image_df = pd.DataFrame({
    'Date': image_dates,
    'Image_Index': range(len(image_dates))
})

# =====================================
# LOAD SEQUENCE DATA
# =====================================
X_sequences = np.load("../data/X_sequences.npy")
y_targets = np.load("../data/y_targets.npy")
sequence_dates = pd.read_csv("../data/sequence_dates.csv")
sequence_dates['Date'] = pd.to_datetime(sequence_dates['Date'])

sequence_df = pd.DataFrame({
    'Date': sequence_dates['Date'],
    'Seq_Index': range(len(sequence_dates))
})

# =====================================
# ALIGN IMAGE + SEQUENCE
# =====================================
merged = pd.merge(image_df, sequence_df, on='Date', how='inner')

print("Final matched samples:", merged.shape)

X_images = processed_images[merged['Image_Index'].values]
X_seq = X_sequences[merged['Seq_Index'].values]
y = y_targets[merged['Seq_Index'].values]

print(X_images.shape)
print(X_seq.shape)
print(y.shape)

# =====================================
# TRAIN TEST SPLIT
# =====================================
X_img_train, X_img_test, X_seq_train, X_seq_test, y_train, y_test = train_test_split(
    X_images, X_seq, y, test_size=0.2, random_state=42
)

# scale target
scaler = MinMaxScaler()
y_train_scaled = scaler.fit_transform(y_train.reshape(-1,1))
y_test_scaled = scaler.transform(y_test.reshape(-1,1))

# =====================================
# CNN BRANCH
# =====================================
img_input = Input(shape=(224,224,3))

x1 = Conv2D(32, (3,3), activation='relu')(img_input)
x1 = MaxPooling2D((2,2))(x1)

x1 = Conv2D(64, (3,3), activation='relu')(x1)
x1 = MaxPooling2D((2,2))(x1)

x1 = Conv2D(128, (3,3), activation='relu')(x1)
x1 = MaxPooling2D((2,2))(x1)

x1 = Flatten()(x1)
x1 = Dense(128, activation='relu')(x1)

# =====================================
# TCN + TRANSFORMER BRANCH
# =====================================
seq_input = Input(shape=(7,11))

x2 = Conv1D(64, kernel_size=2, activation='relu', padding='causal')(seq_input)
x2 = Conv1D(64, kernel_size=2, activation='relu', padding='causal')(x2)

attn = MultiHeadAttention(num_heads=4, key_dim=32)(x2, x2)
x2 = Add()([x2, attn])
x2 = LayerNormalization()(x2)

x2 = GlobalAveragePooling1D()(x2)
x2 = Dense(128, activation='relu')(x2)

# =====================================
# MERGE
# =====================================
combined = Concatenate()([x1, x2])

z = Dense(128, activation='relu')(combined)
z = Dropout(0.3)(z)
z = Dense(64, activation='relu')(z)

output = Dense(1)(z)

model = Model(inputs=[img_input, seq_input], outputs=output)

model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

model.summary()

# =====================================
# TRAIN
# =====================================
history = model.fit(
    [X_img_train, X_seq_train],
    y_train_scaled,
    validation_split=0.2,
    epochs=30,
    batch_size=8
)

# =====================================
# PREDICT
# =====================================
pred_scaled = model.predict([X_img_test, X_seq_test])
pred = scaler.inverse_transform(pred_scaled)

# =====================================
# EVALUATION
# =====================================
mae = mean_absolute_error(y_test, pred)
rmse = np.sqrt(mean_squared_error(y_test, pred))
r2 = r2_score(y_test, pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)

# =====================================
# PLOT
# =====================================
plt.figure(figsize=(8,5))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.legend()
plt.show()
```
