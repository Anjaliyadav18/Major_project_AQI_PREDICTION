```python
import pandas as pd
import numpy as np

# LOAD DATA
df = pd.read_csv("../data/cleaned_aqi_final.csv")
df['Date'] = pd.to_datetime(df['Date'])

# same years
df = df[df['Date'].dt.year.isin([2020, 2021, 2024])]

# sort
df = df.sort_values('Date').reset_index(drop=True)

# FEATURES
feature_cols = [
    'PM2.5', 'PM10', 'NO', 'NO2', 'NOx',
    'NH3', 'CO', 'SO2', 'O3',
    'Benzene', 'Toluene', 'Xylene',
    'AQI'
]

sequence_length = 14

X_sequences = []
y_targets = []
target_dates = []

for i in range(sequence_length, len(df)):
    X_sequences.append(
        df[feature_cols].iloc[i-sequence_length:i].values
    )

    y_targets.append(df['AQI'].iloc[i])
    target_dates.append(df['Date'].iloc[i])

X_sequences = np.array(X_sequences)
y_targets = np.array(y_targets)

print("Data shape:", df.shape)
print("Sequence input shape:", X_sequences.shape)
print("Target shape:", y_targets.shape)

# SAVE
np.save("../data/X_sequences.npy", X_sequences)
np.save("../data/y_targets.npy", y_targets)

sequence_dates = pd.DataFrame({
    'Date': target_dates
})

sequence_dates.to_csv("../data/sequence_dates.csv", index=False)

print("Saved successfully.")
```
